```python
eng = 80
math = 100
urdu = 80

if eng >=80 and urdu >=80:
    print('Masha Allah')
else:
    print('Ok')
```

```python
temperature = 35
is_raining = False
is_traffic = True
is_weekend = True

#                                               FALSE         TRUE
if temperature <= 40 and is_weekend and not is_raining and not is_traffic:
#       TRUE                TRUE                TRUE          FALSE
    print("Great Day")
else:
    print('Bad Day')
```

## NESTED IF ELSE

```python
age = 13
height = 4

if height > 4:
    if age <10:
        print("You can't go")
    elif age>=10 and age<15:
        print("You need to sit in the middle")
    elif age>=15 and age<18:
        print("You will sit in last")
    else:
        print('You will sit in front')
else:
    print("You can't go")
```
