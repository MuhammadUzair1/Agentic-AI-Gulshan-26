## 1. Comparison Operators
These compare two values and always give back a **boolean**: `True` or `False`.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `==` | Equal to | `5 == 5` | `True` |
| `>=` | Greater than **or** equal to | `5 >= 5` | `True` |
| `<=` | Less than **or** equal to | `4 <= 3` | `False` |
| `!=` | **Not** equal to | `5 != 3` | `True` |

> ⚠️ `=` **stores** a value (`a = 5`). `==` **compares** two values (`a == 5`). Don't mix them up!

````python
a = 1
b = 4

a > b
````

````python
print(a < b)
print(a == b)
print(a != b)
print(a >= 1)
print(b <= 3)
````

````python
# Comparisons work on strings too (and they are case-sensitive)
print('uzair' == 'uzair')
print('Uzair' == 'uzair')
````

## 2. `if` / `else`

```python
if condition:
    # runs when the condition is True
else:
    # runs when the condition is False
```

- Don't forget the colon **`:`** at the end of the `if` and `else` lines.
- The code inside **must be indented** (4 spaces, or press `Tab`).

````python
if a > b:
    print(a, 'is greater')
else:
    print(b, 'is greater')
````

### Indentation decides what is *inside* the `if`

````python
if a <= b:
    print(a)         # INSIDE the if  -> runs only when a <= b
print('good')        # NOT part of the if -> always runs
````

## 3. `if` / `elif` / `else`
Use `elif` (short for *else if*) to check **more than two** possibilities.
Python checks the conditions **from top to bottom** and runs **only the first one that is `True`**. It skips everything else.

### ❌ Common mistake: wrong order
Marks = 85 should be *"Masha Allah"*, but we get *"Good"*. Why?
Because `85 > 50` is checked **first**, it's `True`, so Python stops right there!

> 💡 To type your own marks, swap the `marks = 85` line for: `marks = int(input('Enter marks: '))`

````python
marks = 85

if marks > 50:
    print('Good')
elif marks > 60:
    print('Very Good')
elif marks > 80:
    print('Masha Allah')
elif marks > 90:
    print('Kia Baat Hai!')
else:
    print('Fail')
````

### ✅ Correct: check the **highest** condition first

````python
marks = 85

if marks > 90:
    print('Kia Baat Hai!')
elif marks > 80:
    print('Masha Allah')
elif marks > 60:
    print('Very Good')
elif marks > 50:
    print('Good')
else:
    print('Fail')
````

## 5. Short-hand `if` and `if-else`
For very simple conditions, you can write everything on **one line**.

| Type | Syntax |
|---|---|
| Short-hand `if` | `if condition: do_this` |
| Short-hand `if-else` | `do_this if condition else do_that` |
| Short-hand assignment | `x = value1 if condition else value2` |

````python
a = 5
b = 10

if a > b: print('a is greater')   # condition is False, so nothing prints
````

````python
print(a) if a > b else print(b)
````

````python
bigger = a if a > b else b
print(bigger)
````

````python
# Real example
marks = 45
result = 'Pass' if marks >= 50 else 'Fail'
print(result)
````

## 6. Logical Operators: `and`, `or`, `not`
Use these to **combine** conditions.

| Operator | Result is `True` when... | Example |
|---|---|---|
| `and` | **Both** conditions are `True` | `age >= 18 and has_cnic` |
| `or` | **At least one** condition is `True` | `is_weekend or is_holiday` |
| `not` | **Flips** the value: `True` ↔ `False` | `not is_raining` |

> ⚠️ In Python we write the **words** `and`, `or`, `not`. Symbols like `&&`, `||` and `!` from other languages **don't** work here. (`&` and `|` exist in Python, but they are *bitwise* operators, which is a different topic.)

### Truth table
| A | B | `A and B` | `A or B` | `not A` |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

### `and`: both must be true

````python
eng = 80
math = 100
urdu = 80

if eng >= 80 and urdu >= 80:
    print('Masha Allah')
else:
    print('Ok')
````

### `or`: at least one must be true

````python
if eng >= 90 or math >= 90:      # eng is False, math is True -> True
    print('MASHA ALLAH')
````

### `not`: flips the result

````python
print(eng == urdu)          # True  (both are 80)
print(not eng == urdu)      # False (not flips True -> False)

if not eng == urdu:
    print('Not equal')      # does NOT print, because the condition is False
else:
    print('Equal')
````

````python
have_phone = False

if not have_phone:
    print('You need a phone for this course 📱')
````

## 7. ⭐ Operator Precedence: Which Part Runs First?
Just like in maths (**BODMAS**: multiply before you add), Python follows an **order** when a condition has many operators.

### Precedence order (top = runs first)

| Order | Operator | Example |
|---|---|---|
| 1️⃣ | `( )` Brackets | `(a or b)` |
| 2️⃣ | Arithmetic: `*` `/` `%` then `+` `-` | `marks + 5` |
| 3️⃣ | Comparison: `>` `<` `==` `>=` `<=` `!=` | `marks > 50` |
| 4️⃣ | `not` | `not is_raining` |
| 5️⃣ | `and` | `a and b` |
| 6️⃣ | `or` | `a or b` |

> 🧠 **Easy way to remember:** **B**rackets → **M**aths → **C**ompare → **N**OT → **A**ND → **O**R
> *"**B**ig **M**angoes **C**an **N**ever **A**ppear **O**range"* 🥭

### Example 1: `and` runs before `or`
```
True or False and False
        └──── and first ────┘
True or       False
= True
```

````python
print(True or False and False)      # and first  -> True or False  -> True
print((True or False) and False)    # brackets first -> True and False -> False
````

### Example 2: `not` runs before `and`
`not True and False` means `(not True) and False`, **not** `not (True and False)`.

````python
print(not True and False)       # (not True) and False -> False and False -> False
print(not (True and False))     # not (False)          -> True
````

### Example 3: comparison runs before `and` / `or`
Python first turns every comparison into `True`/`False`, then applies `and`/`or`.

````python
marks = 75
attendance = 60

#       True        and       False
print(marks > 50 and attendance >= 75)    # -> False

#       True        or        False
print(marks > 50 or attendance >= 75)     # -> True
````

### Example 4: maths runs before comparison

````python
marks = 48
grace = 5

#  48 + 5 = 53 first, then 53 >= 50
print(marks + grace >= 50)    # True
````

### Example 5: a great day? (step by step)
`not` is applied first, then all the `and`s from left to right.

````python
temperature = 35
is_raining = False
is_traffic = True
is_weekend = True

# Step 1 - comparisons:   temperature <= 40         -> True
# Step 2 - not:           not is_raining            -> True
#                         not is_traffic            -> False
# Step 3 - and:           True and True and True and False -> False
if temperature <= 40 and is_weekend and not is_raining and not is_traffic:
    print('Great Day')
else:
    print('Bad Day')
````

### Example 6: brackets change the answer! 🍕
I want a pizza with **chicken** AND (**olives** OR **mushrooms**).

````python
chicken = False
olive = False
mushrooms = True

# ✅ With brackets: chicken and (olive or mushrooms)
#                   False   and (True)  -> False
if chicken and (olive or mushrooms):
    print('My Pizza 😋')
else:
    print('Not my pizza')
````

````python
# ❌ Without brackets Python reads it as: (chicken and olive) or mushrooms
#                                         (False)            or True -> True
if chicken and olive or mushrooms:
    print('My Pizza 😋  <- wrong! There is no chicken!')
else:
    print('Not my pizza')
````

> ✅ **Golden rule:** When you mix `and` with `or`, **always use brackets**. They make your code correct *and* easier to read.

### Example 7: leap year (a famous precedence example)
A year is a leap year if it's divisible by **4** but **not by 100**, *or* if it's divisible by **400**.

````python
for year in [2024, 2025, 1900, 2000]:
    is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
    print(year, '->', 'Leap year' if is_leap else 'Not a leap year')
````

> The `for` loop above just repeats the check for each year. We'll learn loops in **Class 9**!

## 8. Nested `if`: an `if` inside another `if`
Use it when a second decision only matters **after** the first one is true.

🎢 **Ride rules:** You must be **taller than 4 ft**. Then your seat depends on your age.

````python
age = 13
height = 5

if height > 4:
    if age < 10:
        print("You can't go")
    elif age >= 10 and age < 15:
        print('You need to sit in the middle')
    elif age >= 15 and age < 18:
        print('You will sit in the last row')
    else:
        print('You will sit in front')
else:
    print("You can't go - too short")
````

> 💡 **Tip:** Python lets you chain comparisons. `age >= 10 and age < 15` can be written as `10 <= age < 15`.

````python
age = 16

if 10 <= age < 15:
    print('Middle')
elif 15 <= age < 18:
    print('Last row')
````
