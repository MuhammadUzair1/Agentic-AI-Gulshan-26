## 1. What is an Error?

There are **two kinds** of errors in Python:

| Kind | When it happens | Example | Can `try` catch it? |
|---|---|---|---|
| **Syntax Error** | *Before* the code runs: you broke Python's grammar rules | `print('hi'` (missing bracket) | ❌ No |
| **Exception** | *While* the code runs: the grammar is fine but something goes wrong | `10 / 0` | ✅ Yes |

### Common exceptions you will see

| Exception | Why it happens | Example |
|---|---|---|
| `NameError` | Using a variable that doesn't exist | `print(xyz)` |
| `ZeroDivisionError` | Dividing by zero | `10 / 0` |
| `ValueError` | Right type, but wrong value | `int('abc')` |
| `TypeError` | Wrong type for the operation | `'5' + 5` |
| `IndexError` | List index doesn't exist | `[1, 2, 3][10]` |
| `KeyError` | Dictionary key doesn't exist | `{'a': 1}['b']` |
| `FileNotFoundError` | The file doesn't exist | `open('nofile.txt')` |

Let's see what happens **without** exception handling. The program crashes and the last line **never runs**:

````python
print('Start of program')
print(10 / 0)                   # crash here!
print('End of program')         # this line NEVER runs
````

## 2. `try` / `except`

```python
try:
    # code that MIGHT cause an error
except:
    # runs ONLY if an error happened in the try block
```

- Python **tries** the code inside `try`.
- If an error happens, it **jumps** to `except` instead of crashing.
- If there is no error, the `except` block is **skipped**.

````python
# x was never created, so print(x) causes a NameError
try:
    print(x)
except:
    print('Something went wrong!')

print('The program is still running 🎉')
````

````python
# Converting user input: int('abc') is impossible
user_input = 'abc'          # imagine the user typed this

try:
    age = int(user_input)
    print('Your age is', age)
except:
    print('Please enter a number, not letters!')
````

## 3. Catching Specific Errors

A plain `except:` catches **every** error, so we don't know *what* went wrong.
It's better to name the exact error we expect:

```python
try:
    ...
except ZeroDivisionError:
    ...
```

````python
# NameError
try:
    print(student_name)
except NameError:
    print('VARIABLE NAME NOT FOUND')
````

````python
# ZeroDivisionError
try:
    print(10 / 0)
except ZeroDivisionError:
    print('YOU CANNOT DIVIDE BY ZERO')
````

````python
# IndexError
fruits = ['apple', 'mango', 'banana']

try:
    print(fruits[5])
except IndexError:
    print('There are only', len(fruits), 'fruits in the list')
````

````python
# KeyError
student = {'name': 'Sara', 'age': 20}

try:
    print(student['city'])
except KeyError:
    print("This student has no 'city' key")
````

⚠️ If the error that happens is **different** from the one you named, it is **not caught** and the program still crashes:

````python
try:
    print(10 / 0)           # this is a ZeroDivisionError...
except NameError:           # ...but we only catch NameError
    print('Name not found')
````

## 4. Multiple `except` Blocks

One `try` can have **many** `except` blocks, like `if` / `elif`. Python checks them **from top to bottom** and runs the **first one that matches**.

A plain `except:` at the end catches anything not matched above. It must always be the **last** one.

````python
def divide(a, b):
    try:
        result = a / b
        print(a, '/', b, '=', result)
    except ZeroDivisionError:
        print('Error: cannot divide by zero')
    except TypeError:
        print('Error: please give numbers only')
    except:
        print('Error: something else went wrong')

divide(10, 2)
divide(10, 0)
divide(10, 'five')
````

You can also catch **several** errors in **one** `except` by putting them in a tuple:

````python
for value in ['25', 'abc', None]:
    try:
        print(int(value) + 5)
    except (ValueError, TypeError):
        print('Could not convert', repr(value), 'to a number')
````

## 5. Getting the Error Message with `as e`

`except SomeError as e:` stores the error in the variable `e`, so you can print Python's own message.

`Exception` is the "parent" of almost all errors, so `except Exception as e:` catches (nearly) everything **and** tells you what happened.

````python
try:
    print('5' + 5)
except TypeError as e:
    print('Error message:', e)
````

````python
tests = ['10 / 0', 'int("hello")', '[1, 2][7]']

for test in tests:
    try:
        eval(test)              # eval() runs a string as Python code
    except Exception as e:
        print(type(e).__name__, '->', e)
````

## 6. `else` and `finally`

```python
try:
    # code to try
except:
    # runs if there WAS an error
else:
    # runs if there was NO error
finally:
    # ALWAYS runs, error or no error
```

| Block | Runs when... |
|---|---|
| `try` | Always (first) |
| `except` | There **was** an error |
| `else` | There was **no** error |
| `finally` | **Always**, no matter what |

````python
# No error -> try runs, then ELSE runs
try:
    a = 10
    b = 20
    c = a / b
    print(c)
except ZeroDivisionError:
    print('NOT DIVISIBLE BY ZERO')
except:
    print('WRONG CODE')
else:
    print('No errors! The else block ran.')
````

````python
# Error (d doesn't exist) -> except runs, then FINALLY runs
try:
    a = 10
    b = 20
    c = a / b
    print(d)
except ZeroDivisionError:
    print('NOT DIVISIBLE BY ZERO')
except:
    print('WRONG CODE')
finally:
    print('finally ALWAYS runs')
````

````python
# All four together
def check_division(a, b):
    print(f'--- {a} / {b} ---')
    try:
        result = a / b
    except ZeroDivisionError:
        print('except  : cannot divide by zero')
    else:
        print('else    : the answer is', result)
    finally:
        print('finally : done checking')

check_division(10, 4)
check_division(10, 0)
````

> 💡 **Real-life use of `finally`:** cleaning up, like closing a file, closing a database connection or logging out, which must happen even if something failed.

## 7. Nested `try`

You can put a `try` **inside** another `try`. The inner one handles its own errors; anything it doesn't catch goes to the outer one.

````python
try:
    print('ALPHA')
    try:
        print('BETA')
        print(unknown_variable)         # NameError
    except NameError:
        print('Inner except: WRONG VARIABLE')

    print(10 / 0)                       # ZeroDivisionError
except ZeroDivisionError:
    print('Outer except: cannot divide by zero')
````

## 8. ⚠️ Why `SyntaxError` Can't Be Caught

Python reads (checks) the **whole cell / file first**, before running anything.
If the grammar is broken, it stops **immediately**, so the `try` never even starts.

Notice in the output below: even `'ALPHA'` is **not** printed, and the `except SyntaxError` doesn't help.

````python
try:
    print('ALPHA')
    total = 10 +            # broken grammar: nothing after +
except SyntaxError:
    print('ERROR')
````

> ✅ The fix for syntax errors is simple: **correct the code**. VS Code underlines them in red before you even run the file.

## 9. `raise`: Creating Your Own Errors

Sometimes the code is fine for Python, but **wrong for our program**. For example, an age of `-5`.
We can stop the program on purpose with `raise`:

```python
raise ValueError('Your message here')
```

````python
name = 'UZAIR'

if 'u' in name.lower():
    raise ValueError('Do not use the letter U')
````

Errors we `raise` can be caught with `try` / `except`, just like Python's own errors:

````python
def set_age(age):
    if age < 0:
        raise ValueError('Age cannot be negative')
    if age > 120:
        raise ValueError('Age is too high')
    return age

for age in [25, -5, 200]:
    try:
        print('Age saved:', set_age(age))
    except ValueError as e:
        print('Invalid age', age, '->', e)
````

## 10. Practice Examples

````python
# --- A "safe" number converter ---
def to_number(text):
    try:
        return float(text)
    except ValueError:
        print(f'"{text}" is not a number, using 0 instead')
        return 0

print(to_number('42.5'))
print(to_number('forty'))
````

````python
# --- ATM withdrawal ---
balance = 5000

def withdraw(amount):
    global balance
    try:
        if amount <= 0:
            raise ValueError('Amount must be more than 0')
        if amount > balance:
            raise ValueError(f'Not enough balance. You have only Rs. {balance}')
        balance -= amount
    except ValueError as e:
        print('❌ Transaction failed:', e)
    else:
        print(f'✅ Withdrew Rs. {amount}. Remaining balance: Rs. {balance}')
    finally:
        print('   Thank you for using our ATM\n')

withdraw(2000)
withdraw(10000)
withdraw(-50)
````

````python
# --- Average marks, skipping bad data ---
marks = ['80', '75', 'absent', '90', '', '65']
valid = []

for m in marks:
    try:
        valid.append(int(m))
    except ValueError:
        print(f'Skipping invalid entry: {m!r}')

print('Valid marks:', valid)
print('Average   :', sum(valid) / len(valid))
````

## 📝 Summary

```python
try:
    risky_code()
except ZeroDivisionError:            # a specific error
    ...
except (ValueError, TypeError):      # several errors at once
    ...
except Exception as e:               # anything else + its message
    print(e)
else:
    ...                              # no error happened
finally:
    ...                              # always runs

raise ValueError('message')          # create your own error
```

✅ **Good habits**
- Catch **specific** errors instead of a bare `except:`, so real bugs don't stay hidden.
- Keep the `try` block **small**: only the lines that can actually fail.
- Show the user a **friendly message**, not a scary red traceback.
