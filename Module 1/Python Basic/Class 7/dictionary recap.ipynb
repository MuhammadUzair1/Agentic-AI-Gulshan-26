## DICTIONARY RECAP
Dictionary is a key value pair

key:value (separeted by **:**)

value can be of any data type

```python
mydict = {
    'name':'Uzair',
    'age':24,
    'gender':'male',
    'fav foods':['biryani','pulao','steak']    
}
```

```python
print(mydict)
```

#### NESTED DICTIONARY

```python
mydict_update = {
    'name':'Uzair',
    'age':24,
    'gender':'male',
    'fav foods':['biryani','pulao','steak'],
    'address': {
        'province':'Sindh',
        "city":'Karachi',
        'area':'FB Area'
    }
}
```

```python
mydict_update
```

#### Dictionary Slicing

```python
mydict_update['address']
```

```python
mydict_update['address']['city']
```

```python
type(mydict_update['age'])
```

```python
type(mydict_update['address']['province'])
```

```python
# DICTIONARIES ARE MUTABLE
mydict['age']=89
```

```python
mydict
```

```python
# DICTIONARIES DON'T ALLOW DUPLICATES
mydict = {
    'name':'Uzair',
    'age':24,
    'gender':'male',
    'fav foods':['biryani','pulao','steak'],
    'age':45  
}
mydict
```

```python
# GET ALL THE KEYS
mydict_update.keys()
```

```python
type(mydict_update.keys())
```

```python
list(mydict_update['address'].keys())
```

```python
# GETTING ALL VALUES
mydict_update.values()
```

```python
list(mydict_update.values())
```

```python
# GETTING KEY VALUE PAIRS   
mydict_update.items()
```

```python
# UPDATE A KEY
mydict_update['name']='Huzaifa'
mydict_update
```

```python
# Adding a KEY VLAUE
mydict_update['car name'] = 'audi'
mydict_update
```

```python
# UPDATE
mydict_update.update({'age':30})
mydict_update
```

```python
# REMOVING A VALUE
mydict_update.pop('car name')
mydict_update
```

```python
# ADDING ELEMENT
mydict_update['number'] = '03112025620'
```

```python
# ADDING ELEMENT
mydict_update['education'] = 'Computer Science'
```

```python
# POPITEMS
mydict_update.popitem()
mydict_update
```
