## 2. Dictionary Recap
A dictionary is a collection of **key : value** pairs.

- The key and value are separated by a colon **`:`**
- Pairs are separated by commas **`,`**
- A value can be **any data type**

```python
mydict = {
    'name': 'Uzair',
    'age': 24,
    'gender': 'male',
    'fav foods': ['biryani', 'pulao', 'steak']
}
```

```python
print(mydict)
```

## 3. Nested Dictionary
A dictionary **inside** another dictionary. Here, `address` is itself a dictionary.

```python
mydict_update = {
    'name': 'Uzair',
    'age': 24,
    'gender': 'male',
    'fav foods': ['biryani', 'pulao', 'steak'],
    'address': {
        'province': 'Sindh',
        'city': 'Karachi',
        'area': 'FB Area'
    }
}
```

```python
mydict_update
```

## 6. Dictionaries Don't Allow Duplicate Keys
If a key is repeated, the **last** value replaces the earlier one.

```python
mydict = {
    'name': 'Uzair',
    'age': 24,
    'gender': 'male',
    'fav foods': ['biryani', 'pulao', 'steak'],
    'age': 45
}
mydict
```

## 8. Adding, Updating & Removing Items

| Action | Code |
|---|---|
| Update a value | `d['key'] = new_value` |
| Add a new pair | `d['new key'] = value` |
| Update with a method | `d.update({'key': value})` |
| Remove by key | `d.pop('key')` |
| Remove the last added pair | `d.popitem()` |

```python
# UPDATE A VALUE
mydict_update['name'] = 'Huzaifa'
mydict_update
```

```python
# ADD A NEW KEY-VALUE PAIR
mydict_update['car name'] = 'audi'
mydict_update
```

```python
# UPDATE USING update()
mydict_update.update({'age': 30})
mydict_update
```

```python
# REMOVE A KEY
mydict_update.pop('car name')
mydict_update
```

```python
# ADD MORE ELEMENTS
mydict_update['number'] = '03001234567'
mydict_update['education'] = 'Computer Science'
mydict_update
```

```python
# popitem() -> removes the LAST added pair ('education')
mydict_update.popitem()
mydict_update
```

## 11. Copying a Dictionary

| Code | What it does |
|---|---|
| `b = a` | ❌ **Not a copy.** Both names point to the **same** dictionary |
| `b = a.copy()` | ✅ **Shallow copy.** A new dictionary, but nested items (lists/dicts inside) are still **shared** |
| `b = copy.deepcopy(a)` | ✅ **Deep copy.** Everything is copied, including nested items |

```python
# copy() -> a separate dictionary
mydict = mydict_update.copy()
mydict
```

```python
# = is NOT a copy -> mydict2 and mydict are the SAME dictionary
mydict2 = mydict
mydict2.pop('age')
mydict2
```

```python
# 'age' is gone from mydict too, because it is the same dictionary!
mydict
```

```python
# But mydict_update still has 'age', because copy() made a separate dictionary
mydict_update
```

### ⚠️ Shallow copy and nested dictionaries
`copy()` does **not** copy the dictionaries *inside*. Both copies share the same `address`:

```python
shallow = mydict_update.copy()
shallow['address']['city'] = 'Lahore'
mydict_update['address']['city']   # changed in the original too!
```

```python
# Fix: deepcopy() copies everything, including nested items
import copy

deep = copy.deepcopy(mydict_update)
deep['address']['city'] = 'Islamabad'
print(deep['address']['city'])            # Islamabad
print(mydict_update['address']['city'])   # still Lahore, the original is safe
```

## 12. Renaming a Key
Python has no "rename key" method. Instead, **remove** the old key and **add** its value under a new key.

```python
# Step by step
temp = mydict_update['address']
mydict_update.pop('address')
mydict_update['address2'] = temp
mydict_update
```

```python
# Shortcut: pop() returns the value, so we can do it in ONE line
mydict_update['address'] = mydict_update.pop('address2')
mydict_update
```

## 📝 Key Takeaways
- Dictionaries store **key : value** pairs, and keys must be **unique**.
- Use `d['a']['b']` to reach values inside **nested** dictionaries.
- `keys()`, `values()` and `items()` give you the parts of a dictionary. Wrap them in `list()` to get a normal list.
- `b = a` is **not** a copy. Use `copy()`, or `copy.deepcopy()` for nested data.
