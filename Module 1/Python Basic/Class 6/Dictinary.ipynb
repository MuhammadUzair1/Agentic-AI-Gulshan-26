# Module 1 - Python Basics | Class 6: Dictionaries

A **dictionary** stores data as **key : value** pairs, just like a real dictionary stores *word : meaning*.

```python
car = {'brand': 'toyota', 'model': 'corolla', 'year': 1989}
#        key  :  value
```

Dictionaries are:
- **Ordered**: items keep the order you added them in (Python 3.7+)
- **Mutable**: you can add, change and remove items
- **No duplicate keys**: each key appears only once (values *can* repeat)
- **Any data type** can be a value: strings, numbers, lists, booleans, even other dictionaries

## 1. Creating a Dictionary
Use curly brackets `{ }` with `key: value` pairs separated by commas.

````python
thisdict = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989
}
````

````python
print(thisdict)
````

## 2. Dictionaries vs Sets (Ordered vs Unordered)
A **dictionary** keeps the order you wrote it in.
A **set** (also uses `{ }`, but without keys) is **unordered**, so Python may show the items in a different order.

> ⚠️ Never name a variable `set`, `list`, `dict` or `str`. These are built-in Python names and you would break them.

````python
my_set = {10, 20, 30, 40}
print(my_set)
````

## 3. Duplicate Keys Are Not Allowed
If the same key appears twice, the **last value wins**.

````python
thisdict = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989,
    'year': 2026
}
````

````python
thisdict
````

> 💡 In a notebook, writing just the variable name on the last line of a cell displays it, so no `print()` is needed.

## 4. Dictionary Length
`len()` counts the number of **keys**.

````python
print(len(thisdict))
````

## 5. Values Can Be Any Data Type
Here `colors` holds a **list** and `electric` holds a **boolean**.

````python
thisdict = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989,
    'electric': False,
    'colors': ['red', 'black', 'white']
}
thisdict
````

## 6. Another Way to Create a Dictionary: `dict()`
Write `key=value` without quotes around the keys.

> ⚠️ `year='1989'` (in quotes) would be a **string**, while `year=1989` is an **int**. The same goes for `'False'` vs `False`.

````python
thisdict = dict(brand='toyota', model='corolla', year=1989)
thisdict
````

## 7. Accessing Items

| Way | Example | If the key doesn't exist |
|---|---|---|
| Square brackets | `cars['model']` | ❌ `KeyError` |
| `get()` | `cars.get('model')` | ✅ returns `None` (no error) |

````python
cars = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989
}

x = cars['model']
x
````

````python
y = cars.get('brand')
y
````

````python
# get() is safer: it returns None (or a default you choose) instead of crashing
print(cars.get('price'))
print(cars.get('price', 'Not available'))
# cars['price']   # KeyError: 'price'
````

### Getting all keys, values and items

````python
# keys() -> all the KEYS
cars.keys()
````

````python
# values() -> all the VALUES
cars.values()
````

````python
# items() -> all (key, value) PAIRS as tuples
cars.items()
````

## 8. Adding & Changing Items
- If the key **doesn't exist**, it is **added**.
- If the key **already exists**, its value is **changed**.

````python
# ADD a new key
cars['color'] = 'white'
cars
````

````python
# CHANGE an existing key
cars['year'] = 2018
cars
````

````python
# update() -> change or add one or more keys at once
cars.update({'brand': 'Honda', 'model': 'civic'})
cars
````

## 9. Removing Items

| Method | What it removes |
|---|---|
| `pop('key')` | The item with that key (and returns its value) |
| `popitem()` | The **last** added item |
| `del cars['key']` | The item with that key |
| `clear()` | **Everything**, leaving an empty dictionary `{}` |

````python
cars = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989
}
cars.pop('model')
cars
````

````python
cars = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989
}
cars.popitem()
cars
````

````python
cars = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989
}
del cars['year']
cars
````

````python
cars.clear()
cars
````

## 10. Copying a Dictionary
Just like lists, `new = old` does **not** copy: both names point to the **same** dictionary.
Use `copy()` or `dict()` to make a separate copy.

````python
cars = {
    'brand': 'toyota',
    'model': 'corolla',
    'year': 1989
}
my_car = cars.copy()
my_car['year'] = 2024
print(cars)     # original is unchanged
print(my_car)
````

````python
# dict() also makes a copy
car = dict(cars)
car
````

## 📝 Summary

| Task | Code |
|---|---|
| Create | `d = {'a': 1}` or `dict(a=1)` |
| Read a value | `d['a']` or `d.get('a')` |
| Add / change | `d['b'] = 2` or `d.update({'b': 2})` |
| Remove | `d.pop('a')`, `d.popitem()`, `del d['a']`, `d.clear()` |
| All keys / values / pairs | `d.keys()`, `d.values()`, `d.items()` |
| Count | `len(d)` |
| Copy | `d.copy()` or `dict(d)` |
